```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img= cv2.imread("shapes.jpg")
gray = cv2.cvtColor(img,10) #using color code for grayscale
#blur = cv2.GaussianBlur(gray, (5,5),0) #fuzz out some noise in the image, will need to try remove this for the blankboxes
_, thresh = cv2.threshold(gray, 200,255, cv2.THRESH_BINARY_INV) #sets all pixels with a value above 100 to 255
contours, hierarchy1 = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
rect_contours=[]
for ct in contours:
    if len(ct)==4:
        rect_contours.append(ct)

img2=cv2.drawContours(img,rect_contours,-1, (255,0,255),4)

img_rgb=cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```
